# Real-World Retail Data Science

This notebook explores retail transactions, engineers revenue metrics, analyzes product/customer/region performance, and visualizes business patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/retail_sales.csv')
df['date'] = pd.to_datetime(df['date'])
df['revenue'] = df['quantity'] * df['unit_price'] * (1 - df['discount'])
df.head()
```

## Dataset overview

```python
print(df.shape)
display(df.describe(include='all'))
print('Missing values:', df.isna().sum().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## Monthly revenue trend

```python
monthly = df.assign(month=df['date'].dt.to_period('M').astype(str)).groupby('month', as_index=False)['revenue'].sum()
sns.lineplot(data=monthly, x='month', y='revenue', marker='o')
plt.xticks(rotation=45)
plt.title('Monthly Revenue Trend')
plt.show()
```

## Category and regional performance

```python
category = df.groupby('category')['revenue'].sum().sort_values(ascending=False)
region = df.groupby('region')['revenue'].sum().sort_values(ascending=False)
display(category.to_frame('revenue'))
display(region.to_frame('revenue'))
```

## Product and customer analysis

```python
product = df.groupby('product')['revenue'].sum().sort_values(ascending=False)
customer = df.groupby('customer_id')['revenue'].sum().sort_values(ascending=False)
display(product.head(10).to_frame('revenue'))
display(customer.head(10).to_frame('total_spend'))
```

## Discount and revenue relationship

```python
sns.scatterplot(data=df, x='discount', y='revenue', size='quantity', alpha=0.7, legend=False)
plt.title('Discount vs Revenue')
plt.show()
print(df[['discount', 'quantity', 'unit_price', 'revenue']].corr(numeric_only=True)['revenue'].sort_values(ascending=False))
```
